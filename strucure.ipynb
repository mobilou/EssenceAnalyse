```python
# Prix moyen par an

i = 0
j = 12
k = 0
prix_an = []
while i <= 93:
    passed = True
    while passed:
        dico_12 = dict(list(dico_inverse.items())[k:j])
        prix_an.append(sum(list(dico_12.values()))/12)
        passed = False
        if passed == False:
            j += 12
            k += 12
    i += 12
```

```python
# Taux de croissance par an


i = 0
j = 12
k = 0
taux_an = []

while i <= 93:
    passed = True
    while passed:
        dico_12 = dict(list(dico_inverse.items())[k:j])
        taux_an.append(sum(list(dico_12.values()))/12)
        passed = False
        if passed == False:
            j += 12
            k += 12
    i += 12


print(f"Les prix moyen chaques année : {taux_an}")
liste_taux_an = []

val_ini = 0
val_fin = 1
for i in range(1, len(taux_an)):
    croissancetauxan = ((taux_an[val_fin] - taux_an[val_ini]) / taux_an[val_ini]) * 100
    val_ini += 1
    val_fin += 1
    liste_taux_an.append(croissancetauxan)

print(f"Les taux de croissance par an : {liste_taux_an}")
```

```python
# Tracé de la courbe du taux de croissance par an

annee_de_croissance = [2020, 2021, 2022, 2023, 2024, 2025, 2026]

plt.plot(annee_de_croissance, liste_taux_an, linestyle = "--", marker="o")
plt.xlabel("Années")
plt.ylabel("Taux de croissance par an")
plt.title("Evolution des taux par an")
plt.grid(True)
```

```python
# Taux de croissance par mois

i = 0
j = 12
k = 0

while i <= 93:
    passed = True
    valeur_initial = 0
    valeur_final = 1
    liste2 = []
    while passed:
        dico_12 = dict(list(dico_inverse.items())[k:j])
        liste = list(dico_12.values())
        
        for t in range(1, len(liste)):
            croissancetaux = ((liste[valeur_final] - liste[valeur_initial]) / liste[valeur_initial]) * 100
            valeur_initial += 1
            valeur_final += 1
            liste2.append(croissancetaux)
        passed = False
        if passed == False:
            j += 12
            k += 12
    print(liste2)
    i += 12
    
    
```

```python
# Calcul de la médiane

total = len(liste2019) + len(liste2020) + len(liste2021) + len(liste2022) + len(liste2023) + len(liste2024) + len(liste2025) + len(liste2026)


mediane = (total + 1)/2
print(f"La mediane est à la {mediane} postition")
```

```python
lis = liste2019 + liste2020 + liste2021 + liste2022 + liste2023 + liste2024 + liste2025 + liste2026

# On trie la liste afin de trouver la valeurs correspondant à la médiane
lis.sort()

print(f"La valeure correspondante à la médiane est : {lis[43]}")
```

```python
# Calcul des quartiles

q1 = np.percentile(lis, 25)
q3 = np.percentile(lis, 75)

print(q1)
print(q3)
```

```python
# Calcul de l'interquartile

iqr = q3 - q1
print(iqr)
```

```python
# Définir les seils pour détecter les valeurs aberrantes (règle de Tukey)

seuil_bas = q1 - (1.5 * iqr)
seuil_haut = q3 + (1.5 * iqr)

print(seuil_bas)
print(seuil_haut)
```

```python
valeur_basses = [i for i in lis if i < seuil_bas]
valeur_hautes = [i for i in lis if i > seuil_haut]
valeur_normale = [i for i in lis if seuil_bas <= i <= seuil_haut]
```

```python
print(valeur_hautes)
print(valeur_basses)
print(valeur_normale)
```
