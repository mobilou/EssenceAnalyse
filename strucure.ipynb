```python
for i in range(0, 93):
    dico[df['Date'][i]] = df["Prix"][i].item()
```

```python
dico_inverse = dict(reversed(list(dico.items())))
```

```python
print(dico_inverse)
```

```python
for i in range(0, 12):
    dico_12 = dict(list(dico_inverse.items())[:12])
print(dico_12)
```

```python
i = 0
j = 12
k = 0
while i <= 93:
    passed = True
    while passed:
        dico_12 = dict(list(dico_inverse.items())[k:j])
        print(list(dico_12.values()))
        passed = False
        if passed == False:
            j += 12
            k += 12
    i += 12
```

```python
# Prix moyen par an

i = 0
j = 12
k = 0
while i <= 93:
    passed = True
    while passed:
        dico_12 = dict(list(dico_inverse.items())[k:j])
        print(sum(list(dico_12.values()))/12)
        passed = False
        if passed == False:
            j += 12
            k += 12
    i += 12
```

```python
# Prix moyen par an

i = 0
j = 12
k = 0
while i <= 93:
    passed = True
    while passed:
        dico_12 = dict(list(dico_inverse.items())[k:j])
        liste = list(dico_12.values())
        print(liste)
        passed = False
        if passed == False:
            j += 12
            k += 12
    i += 12
```

```python
# Taux de croissance par mois

i = 0
j = 12
k = 0

while i <= 93:
    passed = True
    valeur_initial = 0
    valeur_final = 1
    liste2 = []
    while passed:
        dico_12 = dict(list(dico_inverse.items())[k:j])
        liste = list(dico_12.values())
        
        for t in range(1, len(liste)):
            croissancetaux = ((liste[valeur_final] - liste[valeur_initial]) / liste[valeur_initial]) * 100
            valeur_initial += 1
            valeur_final += 1
            liste2.append(croissancetaux)
        passed = False
        if passed == False:
            j += 12
            k += 12   
        print(liste2)   
    i += 12
    
    
```
